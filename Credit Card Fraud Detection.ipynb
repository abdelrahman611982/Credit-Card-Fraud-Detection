```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt   
import seaborn as sns
```

```python
df = pd.read_csv("data/creditcard.csv")
# Show all columns
pd.options.display.max_columns = None
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum().sum()
df.duplicated().sum().sum()
print(f"Number of duplicated rows: {df.duplicated().sum()}")
print(f"Number of null values: {df.isna().sum().sum()}")
```

```python
df.drop_duplicates(inplace=True)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))


# First boxplot: 'Amount' by 'Class'
df.boxplot(column='Amount', by='Class', ax=ax[0])
ax[0].set_title('Boxplot of Amount by Class')
ax[0].set_xlabel('Class')
ax[0].set_ylabel('Amount')

# Second boxplot: 'Amount' without grouping by 'Class'
df.boxplot(column="Amount", ax=ax[1])
ax[1].set_title('Boxplot of Amount without grouping')
ax[1].set_xlabel('Class')
ax[1].set_ylabel('Amount')
 
plt.suptitle('Boxplots of Amount and V1 by Class')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Amount'], bins=50, kde=True, log_scale=True)
plt.title('Distribution of Amount')
plt.xlabel('Amount')
plt.ylabel('Frequency') 
plt.show()
```

```python
df_fraud = df[df['Class'] == 1]
df_normal = df[df['Class'] == 0].sample(5000, random_state=42)

pd.plotting.scatter_matrix(df_fraud.iloc[:, :6], figsize=(10,10), diagonal='kde')
plt.suptitle("Fraud Only")
plt.show()
```

```python
# Keep only numeric columns
df_numeric = df.select_dtypes(include=[np.number])

# Keep columns with more than 1 unique value
df_numeric = df_numeric.loc[:, df_numeric.nunique() > 1]

# Reduce number of columns if too many (max 10)
if df_numeric.shape[1] > 10:
    df_numeric = df_numeric.iloc[:, :10]

# Plot scatter matrix
ax = pd.plotting.scatter_matrix(
    df_numeric,
    alpha=0.75,
    figsize=(12, 12),
    diagonal='kde'
)

# Calculate correlations
corrs = df_numeric.corr().values

# Add correlation text to upper triangle
for i, j in zip(*np.triu_indices_from(corrs, k=1)):
    ax[i, j].annotate(
        f'Corr = {corrs[i, j]:.3f}',
        (0.8, 0.2),
        xycoords='axes fraction',
        ha='center',
        va='center',
        size=9
    )

plt.suptitle('Scatter and Density Plot', y=1.02)
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

class_counts = df['Class'].value_counts().sort_index()
labels = ['Normal', 'Fraud']

# Bar chart
ax[1].bar(labels, class_counts)
ax[1].set_title('Class Distribution (Count)')
ax[1].set_ylabel('Number of Transactions')

# Pie chart
ax[0].pie(class_counts, labels=labels, autopct='%1.2f%%', startangle=45, shadow=True)
ax[0].set_title('Class Distribution (Percentage)')

plt.tight_layout()
plt.show()
```
