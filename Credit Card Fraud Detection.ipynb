```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt   
import seaborn as sns
```

```python
df = pd.read_csv('creditcard.csv')
# Show all columns
pd.options.display.max_columns = None
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum().sum()
df.duplicated().sum().sum()
print(f"Number of duplicated rows: {df.duplicated().sum()}")
print(f"Number of null values: {df.isna().sum().sum()}")
```

```python
df.drop_duplicates(inplace=True)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))


# First boxplot: 'Amount' by 'Class'
df.boxplot(column='Amount', by='Class', ax=ax[0])
ax[0].set_title('Boxplot of Amount by Class')
ax[0].set_xlabel('Class')
ax[0].set_ylabel('Amount')

# Second boxplot: 'Amount' without grouping by 'Class'
df.boxplot(column="Amount", ax=ax[1])
ax[1].set_title('Boxplot of Amount without grouping')
ax[1].set_xlabel('Class')
ax[1].set_ylabel('Amount')
 
plt.suptitle('Boxplots of Amount and V1 by Class')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Amount'], bins=50, kde=True, log_scale=True)
plt.title('Distribution of Amount')
plt.xlabel('Amount')
plt.ylabel('Frequency') 
plt.show()
```
